## What is Linear Regression?

**Linear Regression** is a statistical technique used to estimate the linear relationship between a **continuous dependent variable (Y)** and one or more **independent variables (X)**.

- **Independent Variable (X):** Also known as a *predictor* or *feature*. Its values are associated with changes in the dependent variable.
- **Dependent Variable (Y):** Also known as the *outcome*, *response*, or *target variable*. It is the continuous variable that the model aims to estimate or predict.

>### Simple vs. Multiple Linear Regression
>
>- **Simple Linear Regression:** Uses **one independent variable (X)** to estimate one continuous dependent variable (Y).
>- **Multiple Linear Regression:** Uses **two or more independent variables** to estimate one continuous dependent variable (Y).
>

<br>

>### Why Learn Simple Linear Regression?
>
>Simple Linear Regression provides the foundation for understanding more advanced regression models.
>
>The fundamental concepts of **model assumptions, parameter estimation, model construction, evaluation metrics, and interpretation** can be extended to Multiple Linear Regression.
>
>Learning this technique combines several essential Data Science skills:
>
>- **Python Programming:** Building and evaluating regression models.
>- **Exploratory Data Analysis (EDA):** Exploring relationships and checking model assumptions.
>- **Statistics:** Understanding model parameters, assumptions, and evaluation.
>

<br>

**Key Takeaway:** Simple Linear Regression is not only a predictive technique; it also helps us understand relationships between variables and support data-driven business decisions.

## 1. Ordinary Least Squares (OLS)

<br>

### 1.1. The Simple Linear Regression Equation

**Simple Linear Regression** estimates the linear relationship between one independent variable ($X$) and one continuous dependent variable ($Y$).

The estimated regression line is expressed as:

$$\LARGE \boxed{\hat{y}_i = \hat{\beta}_0 + \hat{\beta}_1 x_i}$$

Where:

- $\hat{y}_i$: Predicted value of the dependent variable.
- $x_i$: Observed value of the independent variable.
- $\hat{\beta}_0$: Estimated **intercept** (predicted $Y$ when $X=0$).
- $\hat{\beta}_1$: Estimated **slope** (expected change in predicted $Y$ for a one-unit increase in $X$).

> **NOTE:** The hat symbol ($\hat{}$) indicates an **estimate**. Population parameters ($\beta_0$, $\beta_1$) are unknown, so we estimate them using `sample data`.

<br>

### 1.2. Residuals — Measuring Prediction Errors

A **residual** is the difference between an observed value and its corresponding predicted value.

$$\LARGE \boxed{e_i = y_i - \hat{y}_i}$$

Where:

- $e_i$: Residual for observation $i$.
- $y_i$: Actual or observed value.
- $\hat{y}_i$: Predicted value from the regression model.

>**Residual Interpretation:**
>
>- $e_i > 0$: The model **underestimates** the actual value.
>- $e_i < 0$: The model **overestimates** the actual value.
>- $e_i = 0$: The prediction matches the observation exactly.

>---
> **IMPORTANT:** In OLS regression **with an intercept**, the sum of residuals equals zero (*up to numerical precision*):
>
> $$\LARGE \sum_{i=1}^{n} e_i = 0$$
>
> This property does not necessarily hold for regression models without an intercept or for other estimation methods.
>
>---
>
<br>

### 1.3. Sum of Squared Residuals (SSR)

To measure the model's overall fitting error, OLS uses the **Sum of Squared Residuals (SSR)**:

$$\LARGE \boxed{SSR = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2 = \sum_{i=1}^{n}e_i^2}$$

>**Why square the residuals?**
>
>  1. Prevent positive and negative residuals from canceling each other.
>  2. Penalize larger prediction errors more heavily.
>  3. Provide a mathematical objective that can be minimized to find the best-fitting line.

> **NOTE:** SSR is also commonly called **RSS (Residual Sum of Squares)** or **SSE (Sum of Squared Errors)**. Some statistical references use SSR for *Regression Sum of Squares*, so always verify the notation being used.

<br>

### <span style='color:red'>1.4. Ordinary Least Squares — Finding the Best-Fit Line

**Ordinary Least Squares (OLS)** is a parameter estimation method that finds the regression line minimizing the sum of squared residuals.

Mathematically:

$$\LARGE \boxed{(\hat{\beta}_0,\hat{\beta}_1)=\underset{\beta_0,\beta_1}{\arg\min}\sum_{i=1}^{n}(y_i-\beta_0-\beta_1x_i)^2}$$

For Simple Linear Regression with an intercept, the OLS estimates are:

**Estimated Slope:**

$$\LARGE \boxed{\hat{\beta}_1 =\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sum_{i=1}^{n}(x_i-\bar{x})^2}}$$

**Estimated Intercept:**

$$\LARGE \boxed{\hat{\beta}_0 = \bar{y}-\hat{\beta}_1\bar{x}}$$

Where $\bar{x}$ and $\bar{y}$ represent the sample means.

> **NOTE:** These closed-form formulas require variation in $X$. Python libraries can estimate the coefficients automatically, without manually testing different regression lines.

<br>

### 1.5. Key Takeaways

> - **OLS** estimates regression coefficients by minimizing squared residuals.
>
> - A **residual** measures the vertical difference between an observed point and the fitted regression line.
>
> - **SSR** summarizes the total squared fitting error.
>
> - The estimated **slope** describes how predictions change with $X$.
>
> - The estimated **intercept** determines where the line crosses the $Y$-axis.
>
> - The best-fitting OLS line minimizes SSR, but this **does not automatically guarantee** good predictive performance on unseen data or that regression assumptions are satisfied.

## 2. Correlation and the Intuition Behind Simple Linear Regression

<br>

### 2.1. What is Correlation?

**Correlation** measures how two variables move together. In Simple Linear Regression, we are interested in the **strength and direction of their linear relationship**.

>- **Positive Correlation:** As $X$ increases, $Y$ tends to increase.
>- **Negative Correlation:** As $X$ increases, $Y$ tends to decrease.
>- **Zero Linear Correlation:** There is no linear association between $X$ and $Y$.

<br>

### 2.2. Pearson's Correlation Coefficient ($r$)

**Pearson's Correlation Coefficient ($r$)** quantifies the strength and direction of the linear relationship between two numerical variables.

$$\LARGE \boxed{-1 \leq r \leq 1}$$

| Correlation ($r$) | Interpretation |
|---|---|
| $r = +1$ | Perfect positive linear correlation |
| $0 < r < 1$ | Positive linear correlation |
| $r = 0$ | No linear correlation |
| $-1 < r < 0$ | Negative linear correlation |
| $r = -1$ | Perfect negative linear correlation |

>**Key Properties:**
>
>  - The **sign** of $r$ indicates the direction of the relationship.
>  - The **absolute value** $|r|$ indicates the strength of the linear relationship.
>  - Values closer to $-1$ or $+1$ indicate stronger linear relationships.
>  - <span style='color:gray'>The correlation coefficient is **unitless**.
>  - Correlation does **not** measure the magnitude of the regression slope.
>
>---
>
> **IMPORTANT:** A correlation coefficient close to zero does not necessarily mean that two variables are unrelated. They may have a strong **nonlinear relationship**, such as $Y=X^2$.
>
>---

<br> 

### <span style='color:red'>2.3. Calculating Pearson's Correlation Coefficient

Pearson's correlation coefficient is calculated as:

$$\LARGE \boxed{r = \frac{\operatorname{Cov}(X,Y)}{SD_X \cdot SD_Y}}$$

Where:

  - $\operatorname{Cov}(X,Y)$: Covariance between $X$ and $Y$.
  - $SD_X$: Standard deviation of $X$.
  - $SD_Y$: Standard deviation of $Y$.

>---
>
>**Covariance** measures how two variables vary together relative to their respective means.
>
>---

For a population:

$$\LARGE \operatorname{Cov}(X,Y) = \frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{n}$$

>---
>
>**NOTE: For a `sample`, covariance and standard deviation calculations use <span style='color:red'>$\LARGE n-1$</span> instead of "<span style='color:orange'>$\LARGE n$</span>".**
>
>---

<BR>

>**Interpretation:**
>
>  - **Positive Covariance:** Both variables tend to move in the same direction relative to their means.
>  - **Negative Covariance:** The variables tend to move in opposite directions.
>  - **Zero Covariance:** No linear association is detected.
>
>
The product of the standard deviations normalizes covariance, producing a unitless correlation coefficient between $-1$ and $+1$.

> **NOTE:** `Correlation is not causation`. A strong association between two variables does not establish that one variable causes changes in the other.

<br>

---

### <span style='color:red'>2.4. Correlation Using Standardized Values

Pearson's correlation can also be calculated using **standardized values (`Z-scores`)**.

$$\LARGE z_{x_i} = \frac{x_i-\bar{x}}{SD_X}\qquad z_{y_i} = \frac{y_i-\bar{y}}{SD_Y}$$

Using population standard deviations:

$$\LARGE \boxed{r = \frac{1}{n}\sum_{i=1}^{n} z_{x_i}z_{y_i}}$$

**Intuition:**

- If both standardized values have the same sign, their product is positive.
- If they have opposite signs, their product is negative.
- Averaging these products measures how consistently the variables move together.

>---
>
> **NOTE:** When using sample standard deviations, the corresponding standardized-product formula uses $1/(n-1)$.
>
>---

<br>

### 2.5. Example — Study Hours and Exam Scores

Consider five students and their exam scores:

| Study Hours ($X$) | Exam Score ($Y$) | $Z_X$ | $Z_Y$ | $Z_X \cdot Z_Y$ |
|---:|---:|---:|---:|---:|
| 2 | 75 | -1.5 | -0.5 | 0.75 |
| 4 | 65 | -0.5 | -1.5 | 0.75 |
| 5 | 80 | 0.0 | 0.0 | 0.00 |
| 6 | 95 | 0.5 | 1.5 | 0.75 |
| 8 | 85 | 1.5 | 0.5 | 0.75 |

**Summary Statistics:**

- Mean of $X$: $\bar{x}=5$
- Mean of $Y$: $\bar{y}=80$
- Population standard deviation of $X$: $SD_X=2$
- Population standard deviation of $Y$: $SD_Y=10$

Calculating correlation:

$$
r = \frac{0.75+0.75+0+0.75+0.75}{5}
$$

$$
\boxed{r=0.6}
$$

**Interpretation:** Study hours and exam scores have a positive linear correlation. Students who study more hours tend to achieve higher exam scores, although the relationship is not perfectly linear.

<br>

<br>

<br>

### 2.6. The Intuition Behind Linear Regression

**Linear Regression** estimates the expected or average value of $\LARGE Y$ for a given value of $\LARGE X$, assuming a linear relationship.

- Without information about $\LARGE X$, the overall mean of $\LARGE Y$ is a natural baseline prediction when minimizing squared errors.

- When $\LARGE X$ is available, we can use it to improve predictions by estimating how the average value of $\LARGE Y$ changes with $\LARGE X$.

The regression line represents the **best-fitting linear approximation** of the relationship between $\LARGE X$ and $\LARGE Y$.

<br>

### 2.7. <span style='color:red'>Connecting Correlation to the Regression Equation

The Simple Linear Regression equation is:

$$\LARGE \boxed{\hat{y}=\hat{\beta}_0+\hat{\beta}_1 x}$$

Two important properties connect Pearson's correlation to the OLS regression line.

<br>

**`Property 1 — The Regression Line Passes Through the Means`**

The point $\LARGE (\bar{x},\bar{y})$ always lies on the OLS regression line when an intercept is included.

$$\LARGE \boxed{\bar{y}=\hat{\beta}_0+\hat{\beta}_1\bar{x}}$$

<br>

**`Property 2 — Correlation Determines the Standardized Slope`**

For every increase of **one standard deviation in $\LARGE X$**, the predicted value of $\LARGE Y$ changes by **$\LARGE r$ standard deviations**, on average.

Therefore, the estimated slope is:

$$\LARGE \boxed{\hat{\beta}_1=r\frac{SD_Y}{SD_X}}$$

The estimated intercept is:

$$\LARGE \boxed{\hat{\beta}_0=\bar{y}-\hat{\beta}_1\bar{x}}$$

>---
>
> **IMPORTANT:** Correlation determines the sign and standardized magnitude of the slope, but the actual slope also depends on the ratio of the variables' standard deviations.
>
>---

<br>


### 2.8. Example — Deriving the Regression Equation

Using the previous example:

| Statistic | Value |
|---|---:|
| Mean of $X$ | 5 |
| Mean of $Y$ | 80 |
| $SD_X$ | 2 |
| $SD_Y$ | 10 |
| Pearson's $r$ | 0.6 |

---

**Step 1 — `Calculate the Slope`**

$$\LARGE \hat{\beta}_1=r\frac{SD_Y}{SD_X}$$

$$\LARGE \hat{\beta}_1=0.6\left(\frac{10}{2}\right)=3$$

---


**Step 2 — `Calculate the Intercept`**

$$\LARGE \hat{\beta}_0=\bar{y}-\hat{\beta}_1\bar{x}$$

$$\LARGE \hat{\beta}_0=80-3(5)=65$$

---

**Step 3 — `Construct the Regression Equation`**

$$\LARGE \boxed{\hat{y}=65+3x}$$


>**Interpretation:**
>
>  - **Slope = 3:** Each additional hour of studying is associated with an estimated increase of 3 points in the exam score.
>
>  - **Intercept = 65:** The model predicts an exam score of 65 when study hours equal zero. This is an extrapolation beyond the observed study-hour range and should be interpreted cautiously.

<br>

---

**Step 4 — `Make a Prediction`**

For a student who studies 7 hours:

$$\LARGE \hat{y}=65+3(7)=86 $$

**Predicted Exam Score: 86 points.**

<br>

### <span style='color:red'>2.9. Key Takeaways

> - Pearson's $r$ measures the **direction and strength of linear association**.
>
> - The closer $\LARGE |r|$ is to 1, the stronger the linear relationship.
>
> - Correlation does not describe the magnitude of the regression slope.
>
> - A low correlation does not rule out a nonlinear relationship.
>
> - Standardization makes correlation independent of measurement units.
>
> ---
> 
> - Linear Regression estimates the average response of $Y$ given $X$, under a linear model.
>
> - The OLS regression line with an intercept passes through $(\bar{x},\bar{y})$.
>
> - Pearson's $r$ helps calculate the regression slope.
>
> - The regression slope and correlation coefficient have the same sign.
>
> - **Correlation measures association; regression provides an equation for prediction.**
>
> - A regression relationship does not necessarily imply causation.

## 3. Linear Regression Assumptions

<br>

### 3.1. Introduction — Why Are Regression Assumptions Important?

**Linear Regression Assumptions** are statistical conditions that help determine whether a linear regression model is appropriate for a particular dataset and whether its results can be interpreted reliably.

Before constructing and interpreting a regression model, we must verify that its underlying assumptions are reasonably satisfied.

Within the **PACE Framework**, checking assumptions connects the **Analyze** and **Construct** phases.

| PACE Phase | Main Activity |
|---|---|
| **Plan** | Define the business problem, objectives, and data requirements. |
| **Analyze** | Explore the dataset, visualize relationships, and investigate assumptions that can be assessed before fitting. |
| **Construct** | Build the regression model and evaluate assumptions involving residuals. |
| **Execute** | Interpret, communicate, and apply the results while acknowledging limitations. |

>---
>
>**Important:** Not every assumption can be assessed before fitting the model. **Assumptions concerning residuals generally require a fitted model**.
>
>---

<br>

### 3.2. The Four Key Assumptions

The four assumptions are:

1. **Linearity:** The expected relationship between the independent and dependent variables is linear.

2. **Normality:** The model errors are normally distributed, conditional on the predictors.

3. **Independence:** Observations and their associated errors should not exhibit inappropriate dependence.

4. **Homoscedasticity:** The error variance is constant across predictor values.

| Assumption | Main Question | Diagnostic Tool |
|---|---|---|
| **Linearity** | Does a straight line reasonably represent the relationship between $X$ and $Y$? | Scatter plot (*residuals vs. fitted values*) |
| **Normality** | Are the residuals approximately normally distributed? | Q-Q plot; histogram |
| **Independence** | Are observations and errors independent, or at least uncorrelated where required? | Data collection context; residuals vs. observation order |
| **Homoscedasticity** | Does the residual spread remain approximately constant? | Scatterplot (*Residuals vs. fitted values*) |

<br>

>---
>
> **Key Concept:** A regression line can be fitted to almost any numerical dataset, but that does not mean linear regression is necessarily an appropriate statistical model.
>
>---

<br><br>

---

### <span style='color:red'>3.3. Assumption #1 — Linearity

**Linearity** assumes that the conditional mean of the dependent variable $Y$ is a linear function of the independent variable $X$.

For Simple Linear Regression:

$$ \LARGE \boxed{Y_i = \beta_0+\beta_1x_i}$$

>Where:
>
>- $Y_i$: Expected value of $Y$ given $x_i$.
>- $\beta_0$: Population intercept.
>- $\beta_1$: Population slope.
>- $x_i$: Value of the independent variable.

<br>

**How to check linearity:**

Create a scatter plot of $X$ versus $Y$.

- **Assumption reasonably satisfied:** Data points tend to cluster around a straight-line trend.

- **Possible violation:** The relationship displays systematic curvature or another nonlinear pattern.

- **Weak association:** A random-looking cloud may indicate little linear predictive signal, but it does not automatically prove a violation of linearity.

<br>

**Example:**

- Suppose we want to predict an employee's salary based on years of experience.

- If salaries increase at an approximately constant average rate as experience increases, a linear model may be appropriate.

- If salaries initially increase rapidly but later level off, a nonlinear model may provide a better representation.

<br> 

**Additional Statistical Insight:**

- Linearity does not mean that every observed data point must fall directly on the regression line.

- The line represents the **conditional expected value** of $Y$, not an exact deterministic relationship.

<br><br>

---

### <span style='color:red'>3.4. Assumption #2 — Normality of Errors

**Normality** assumes that regression errors follow a normal distribution, conditional on the predictors.

The regression model can be written as:

$$ \LARGE \boxed{Y_i=\beta_0+\beta_1X_i+\varepsilon_i}$$

Under the conventional normal-error model 

- *Epsilon sub i, given X sub i, follows a normal distribution with mean zero and variance sigma squared.*:

$$
\LARGE \boxed{\varepsilon_i \mid X_i \sim N(0,\sigma^2)}
$$


Where:

- $\varepsilon_i$: The unobserved error term.
- $N$: Normal distribution.
- $0$: Mean Zero (*Expected error value*).
- $\sigma^2$: Constant error variance.

<br>

**Important distinction:**

The assumption concerns the **errors**, not the distribution of the independent variable $X$ or the dependent variable $Y$.

Because true errors are unobservable, we examine the **residuals** after constructing the regression model.

The residual for observation $i$ is:

$$\LARGE \boxed{e_i=y_i-\hat{y}_i}$$

<br>

`Checking Normality with a Q-Q Plot`

A **Quantile-Quantile (Q-Q) Plot** compares the quantiles of the observed residual distribution with the theoretical quantiles of a normal distribution.

**Interpretation:**

| Q-Q Plot Pattern | Interpretation |
|---|---|
| <span style='color:green'>Points approximately follow a diagonal reference line | <span style='color:blue'>Residuals are reasonably consistent with normality. |
| Points display systematic curvature | Possible skewness or other departures from normality. |
| Points deviate strongly at the ends | Possible heavy tails, light tails, or extreme observations. |
| Many points depart from the reference line | Normality may be questionable. |

<br>

**Additional Statistical Insight:**

- A Q-Q plot provides visual evidence, not mathematical proof of normality.

- Also, normality is **not required for OLS coefficient estimation itself**. With appropriate conditions, OLS can still provide unbiased or consistent estimates without normally distributed errors.

- Normality is especially relevant to exact small-sample $t$ tests, $F$ tests, and confidence intervals under the classical normal-error model.

<br><br>

---

### <span style='color:red'>3.5. Assumption #3 — Independence of Observations and Errors

The third assumption concerns whether observations and their associated errors exhibit dependence. 

For the regression model, an important related condition is the absence of correlation between errors from different observations.

For distinct observations $i$ and $j$:

$$\LARGE \boxed{\operatorname{Cov}(\varepsilon_i,\varepsilon_j \mid X)=0,\quad i\neq j}$$

Here, $X$ represents the observed predictor design.

This equation expresses **zero conditional error covariance**, which is weaker than full statistical independence.

<br> 

**Why is independence important?**

- If errors are correlated, conventional OLS standard errors may be incorrect, making statistical inference unreliable.

<br>

`Common Situations Where Dependence Can Occur`

- **Time series:** Today's observation may be related to yesterday's observation.

- **Repeated measurements:** Multiple measurements from the same person may be correlated.

- **Grouped observations:** Individuals from the same household, school, or organization may share unobserved influences.

- **Sequential measurements:** Consecutive measurements from a physical system may be correlated.


<br>

Additional diagnostics include:

1. Plot residuals against time or observation order.

2. Examine the autocorrelation function (ACF).

3. Apply the `Durbin-Watson` statistic when appropriate for detecting first-order serial correlation.

---

The `Durbin-Watson statistic` is:

$$
\LARGE \boxed{DW=\frac{\sum_{t=2}^{n}(e_t-e_{t-1})^2}{\sum_{t=1}^{n}e_t^2}}
$$

**Approximate interpretation:**

| Durbin-Watson Value | Interpretation |
|---|---|
| Between 1.5 - 2.5 | Little evidence of first-order autocorrelation. |
| Below 1.5 | Suggests positive first-order autocorrelation. |
| Above 2.5 | Suggests negative first-order autocorrelation. |

The statistic ranges approximately from 0 to 4.

> **Important:** The Durbin-Watson statistic focuses on `first-order serial correlation`. It does not detect every type of dependence, and its interpretation depends on model conditions.

<br>


<br><br>

---

### <span style='color:red'>3.6. Assumption #4 — Homoscedasticity

**Homoscedasticity** means that the variance of the regression errors remains constant across the range of predictor values.

Mathematically:

$$
\LARGE \boxed{\operatorname{Var}(\varepsilon_i \mid X_i=x_i)=\sigma^2}
$$

Where:

- $\operatorname{Var}$: Variance.
- $\varepsilon_i$: Error term.
- $\sigma^2$: Constant error variance.

The opposite condition is called **Heteroscedasticity**, meaning that error variance changes across the predictor range.

<br>

`How to Check Homoscedasticity`

Construct a **Residuals vs. Fitted Values Plot**:

- Horizontal axis: Predicted values $\hat{y}_i$.
- Vertical axis: Residuals $e_i$.

**Interpretation:**

| Residual Pattern | Interpretation |
|---|---|
| Random cloud with approximately constant vertical spread | Consistent with homoscedasticity. |
| Funnel or cone-shaped pattern | Suggests heteroscedasticity. |
| Increasing vertical spread | Error variance may increase with fitted values. |
| Decreasing vertical spread | Error variance may decrease with fitted values. |

<br>


**Additional Statistical Insight:**

- Heteroscedasticity does not automatically make OLS coefficients biased.

- However, it can invalidate conventional homoscedasticity-based standard errors and confidence intervals.

>---
>
>Possible remedies include:
>
> - Heteroscedasticity-robust standard errors.
>
> - Transformations of the response variable, when appropriate.
>
> - Weighted Least Squares (WLS), when the variance structure can be reasonably modeled.
>
> - Improving model specification.
>
>---
>

The following example demonstrates how to create useful diagnostic plots after fitting a Simple Linear Regression model.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.graphics.gofplots import qqplot
from statsmodels.stats.stattools import durbin_watson
```

```python
# Generate reproducible example data
rng = np.random.default_rng(42)

n = 100
X = rng.uniform(1, 10, n)
errors = rng.normal(0, 3, n)
Y = 10 + 2.5 * X + errors

# --------------------------------------------------
# Fit Simple Linear Regression
X_with_constant = sm.add_constant(X)
model = sm.OLS(Y, X_with_constant).fit()

# Extract predictions and residuals
fitted_values = model.fittedvalues
residuals = model.resid

# --------------------------------------------------
# Create diagnostic plots
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# 1. Linearity
sns.regplot(x=X, y=Y, ax=axes[0, 0],
            scatter_kws={"alpha": 0.6})
axes[0, 0].set_title("Linearity: X vs. Y", fontsize=15, fontweight='bold')
axes[0, 0].set_xlabel("Independent Variable (X)")
axes[0, 0].set_ylabel("Dependent Variable (Y)")

# 2. Normality
qqplot(residuals, line="45", fit=True, ax=axes[0, 1])
axes[0, 1].set_title("Normality: Q-Q Plot", fontsize=15, fontweight='bold')

# 3. Independence (requires meaningful observation order)
axes[1, 0].scatter(np.arange(n), residuals, alpha=0.6)
axes[1, 0].axhline(0, color="red", linestyle="--")
axes[1, 0].set_title("Independence: Residuals vs. Observation Order", fontsize=15, fontweight='bold')
axes[1, 0].set_xlabel("Observation Index")
axes[1, 0].set_ylabel("Residuals")

# 4. Homoscedasticity
sns.scatterplot(x=fitted_values, y=residuals,
                ax=axes[1, 1])
axes[1, 1].axhline(0, color="red", linestyle="--")
axes[1, 1].set_title("Homoscedasticity: Residuals vs. Fitted", fontsize=15, fontweight='bold')
axes[1, 1].set_xlabel("Fitted Values")
axes[1, 1].set_ylabel("Residuals")

plt.tight_layout()
plt.show()

# --------------------------------------------------
# Durbin-Watson statistic
print()
print('-'*60)
print("Durbin-Watson:", round(durbin_watson(residuals), 3))
print('-'*60)
```

**Important Notes:**

- This dataset is simulated with linearity, normally distributed independent errors, and constant error variance.

- The four diagnostic plots should generally be consistent with those conditions, although random samples can show apparent deviations.

- Because the simulated observations are generated independently, their array order is not inherently chronological.

- For real datasets, independence diagnostics should use the actual time or collection order where relevant.

- Residual plots are diagnostic tools; they do not guarantee that assumptions are satisfied.

### 3.8. Summary — Comparison of Regression Assumptions

| Assumption | Statistical Requirement | Expected Visual Pattern | Potential Consequence of Violation |
|---|---|---|---|
| **Linearity** | $E[Y\mid X]=\beta_0+\beta_1X$ | Approximately linear trend; no systematic residual curvature | Incorrect mean specification and potentially biased predictions. |
| **Normality** | $\varepsilon\mid X\sim N(0,\sigma^2)$ under the normal-error model | Residual Q-Q points near diagonal | Exact small-sample parametric inference may be unreliable. |
| **Independence** | Errors independent, or at least appropriately uncorrelated | No meaningful serial pattern in ordered residuals | Conventional standard errors and inference may be unreliable. |
| **Homoscedasticity** | $\operatorname{Var}(\varepsilon\mid X)=\sigma^2$ | Approximately constant vertical residual spread | Conventional standard errors may be incorrect; OLS may lose efficiency. |

<br>

### <span style='color:red'>3.9. Key Takeaways

> - **Linearity:** The expected relationship between $X$ and $Y$ should be approximately linear.
>
> - **Normality:** The errors, rather than the original variables, are assumed normally distributed under the classical normal-error model.
>
> - **Independence:** Understand how observations were collected and investigate potential correlation between errors.
>
> - **Homoscedasticity:** The variance of the errors should remain approximately constant.
>
> ---
>
> - A **scatter plot of $X$ vs. $Y$** is useful for examining the linear relationship.
>
> - A **Q-Q plot** helps assess whether residuals are approximately normally distributed.
>
> - **Residuals vs. observation order** can reveal serial patterns.
>
> - **Residuals vs. fitted values** can reveal nonlinear patterns and changes in error variance.
>
> - Violations do not all affect regression in the same way. Some threaten model specification, while others primarily affect statistical inference.
>
> - Regression diagnostics are an **iterative process**, not a one-time checklist.
