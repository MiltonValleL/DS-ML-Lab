## 1. What is Linear Regression?

**Linear Regression** is a statistical technique used to estimate the linear relationship between a **continuous dependent variable (Y)** and one or more **independent variables (X)**.

- **Independent Variable (X):** Also known as a *predictor* or *feature*. Its values are associated with changes in the dependent variable.
- **Dependent Variable (Y):** Also known as the *outcome*, *response*, or *target variable*. It is the continuous variable that the model aims to estimate or predict.

>### Simple vs. Multiple Linear Regression
>
>- **Simple Linear Regression:** Uses **one independent variable (X)** to estimate one continuous dependent variable (Y).
>- **Multiple Linear Regression:** Uses **two or more independent variables** to estimate one continuous dependent variable (Y).
>

<br>

>### Why Learn Simple Linear Regression?
>
>Simple Linear Regression provides the foundation for understanding more advanced regression models.
>
>The fundamental concepts of **model assumptions, parameter estimation, model construction, evaluation metrics, and interpretation** can be extended to Multiple Linear Regression.
>
>Learning this technique combines several essential Data Science skills:
>
>- **Python Programming:** Building and evaluating regression models.
>- **Exploratory Data Analysis (EDA):** Exploring relationships and checking model assumptions.
>- **Statistics:** Understanding model parameters, assumptions, and evaluation.
>

<br>

**Key Takeaway:** Simple Linear Regression is not only a predictive technique; it also helps us understand relationships between variables and support data-driven business decisions.

## 2. Learning Objectives — Simple Linear Regression

Throughout this lab, we will learn how to **set up, construct, evaluate, and interpret** a Simple Linear Regression model using the **PACE Framework**.

<br>

### PACE Framework

**PACE = Plan → Analyze → Construct → Execute**

| Phase | Main Activities |
|---|---|
| **1. Plan** | Review the linear regression equation, understand parameter estimation using Ordinary Least Squares (OLS), and identify the four key regression assumptions. |
| **2. Analyze** | Use Python and Exploratory Data Analysis (EDA) to examine the dataset and verify whether the regression assumptions are satisfied. |
| **3. Construct** | Build a Simple Linear Regression model in Python and evaluate its performance using regression evaluation metrics. |
| **4. Execute** | Interpret the model results and communicate findings to stakeholders and non-technical audiences. |

> **Note:** PACE is an iterative framework. Some phases or activities may need to be revisited during the regression modeling process.

<br>

### Main Topics Covered

1. Linear Regression Equation.
2. Parameter Estimation using **Ordinary Least Squares (OLS)**.
3. The **Four Key Assumptions** of Simple Linear Regression.
4. Assumption Verification using **Python and EDA**.
5. Model Construction in Python.
6. Model Evaluation Metrics.
7. Model Interpretation and Communication.

<br>

### Final Takeaway

The goal of this lab is to develop the ability to build a statistically sound Simple Linear Regression model, evaluate its performance, and translate its results into meaningful insights for business decision-making.
