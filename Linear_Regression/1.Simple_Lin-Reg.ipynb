## 1. What is Linear Regression?

**Linear Regression** is a statistical technique used to estimate the linear relationship between a **continuous dependent variable (Y)** and one or more **independent variables (X)**.

- **Independent Variable (X):** Also known as a *predictor* or *feature*. Its values are associated with changes in the dependent variable.
- **Dependent Variable (Y):** Also known as the *outcome*, *response*, or *target variable*. It is the continuous variable that the model aims to estimate or predict.

>### Simple vs. Multiple Linear Regression
>
>- **Simple Linear Regression:** Uses **one independent variable (X)** to estimate one continuous dependent variable (Y).
>- **Multiple Linear Regression:** Uses **two or more independent variables** to estimate one continuous dependent variable (Y).
>

<br>

>### Why Learn Simple Linear Regression?
>
>Simple Linear Regression provides the foundation for understanding more advanced regression models.
>
>The fundamental concepts of **model assumptions, parameter estimation, model construction, evaluation metrics, and interpretation** can be extended to Multiple Linear Regression.
>
>Learning this technique combines several essential Data Science skills:
>
>- **Python Programming:** Building and evaluating regression models.
>- **Exploratory Data Analysis (EDA):** Exploring relationships and checking model assumptions.
>- **Statistics:** Understanding model parameters, assumptions, and evaluation.
>

<br>

**Key Takeaway:** Simple Linear Regression is not only a predictive technique; it also helps us understand relationships between variables and support data-driven business decisions.

## 3. Ordinary Least Squares (OLS)

<br>

### 3.1. The Simple Linear Regression Equation

**Simple Linear Regression** estimates the linear relationship between one independent variable ($X$) and one continuous dependent variable ($Y$).

The estimated regression line is expressed as:

$$\LARGE \boxed{\hat{y}_i = \hat{\beta}_0 + \hat{\beta}_1 x_i}$$

Where:

- $\hat{y}_i$: Predicted value of the dependent variable.
- $x_i$: Observed value of the independent variable.
- $\hat{\beta}_0$: Estimated **intercept** (predicted $Y$ when $X=0$).
- $\hat{\beta}_1$: Estimated **slope** (expected change in predicted $Y$ for a one-unit increase in $X$).

> **NOTE:** The hat symbol ($\hat{}$) indicates an **estimate**. Population parameters ($\beta_0$, $\beta_1$) are unknown, so we estimate them using sample data.

<br>

### 3.2. Residuals — Measuring Prediction Errors

A **residual** is the difference between an observed value and its corresponding predicted value.

$$\LARGE \boxed{e_i = y_i - \hat{y}_i}$$

Where:

- $e_i$: Residual for observation $i$.
- $y_i$: Actual or observed value.
- $\hat{y}_i$: Predicted value from the regression model.

>**Residual Interpretation:**
>
>- $e_i > 0$: The model **underestimates** the actual value.
>- $e_i < 0$: The model **overestimates** the actual value.
>- $e_i = 0$: The prediction matches the observation exactly.

>---
> **IMPORTANT:** In OLS regression **with an intercept**, the sum of residuals equals zero (up to numerical precision):
>
> $$\LARGE \sum_{i=1}^{n} e_i = 0$$
>
> This property does not necessarily hold for regression models without an intercept or for other estimation methods.
>
>---
>
<br>

### 3.3. Sum of Squared Residuals (SSR)

To measure the model's overall fitting error, OLS uses the **Sum of Squared Residuals (SSR)**:

$$\LARGE \boxed{SSR = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2 = \sum_{i=1}^{n}e_i^2}$$

>**Why square the residuals?**
>
>  1. Prevent positive and negative residuals from canceling each other.
>  2. Penalize larger prediction errors more heavily.
>  3. Provide a mathematical objective that can be minimized to find the best-fitting line.

> **NOTE:** SSR is also commonly called **RSS (Residual Sum of Squares)** or **SSE (Sum of Squared Errors)**. Some statistical references use SSR for *Regression Sum of Squares*, so always verify the notation being used.

<br>

### <span style='color:red'>3.4. Ordinary Least Squares — Finding the Best-Fit Line

**Ordinary Least Squares (OLS)** is a parameter estimation method that finds the regression line minimizing the sum of squared residuals.

Mathematically:

$$\LARGE \boxed{(\hat{\beta}_0,\hat{\beta}_1)=\underset{\beta_0,\beta_1}{\arg\min}\sum_{i=1}^{n}(y_i-\beta_0-\beta_1x_i)^2}$$

For Simple Linear Regression with an intercept, the OLS estimates are:

**Estimated Slope:**

$$\LARGE \boxed{\hat{\beta}_1 =\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sum_{i=1}^{n}(x_i-\bar{x})^2}}$$

**Estimated Intercept:**

$$\LARGE \boxed{\hat{\beta}_0 = \bar{y}-\hat{\beta}_1\bar{x}}$$

Where $\bar{x}$ and $\bar{y}$ represent the sample means.

> **NOTE:** These closed-form formulas require variation in $X$. Python libraries can estimate the coefficients automatically, without manually testing different regression lines.

<br>

### 3.5. Example — Comparing Regression Lines

The lesson illustrates how different slopes and intercepts produce different regression lines.

| Model | Intercept | Slope | Equation |
|---|---:|---:|---|
| Attempt 1 | 2.5 | 1.0 | $\hat{y}=2.5+1.0x$ |
| Attempt 2 | 3.0 | 1.25 | $\hat{y}=3.0+1.25x$ |
| OLS Best Fit | 1.5 | 3.2 | $\hat{y}=1.5+3.2x$ |

**Key Idea:** Rather than selecting a regression line visually or through trial and error, OLS mathematically identifies the coefficients that minimize the total squared residuals for the given dataset.


<br>

### 3.6. Key Takeaways

- **OLS** estimates regression coefficients by minimizing squared residuals.
- A **residual** measures the vertical difference between an observed point and the fitted regression line.
- **SSR** summarizes the total squared fitting error.
- The estimated **slope** describes how predictions change with $X$.
- The estimated **intercept** determines where the line crosses the $Y$-axis.
- The best-fitting OLS line minimizes SSR, but this **does not automatically guarantee** good predictive performance on unseen data or that regression assumptions are satisfied.
